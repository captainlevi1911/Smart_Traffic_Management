## Raw File Ordering

Inspection of consecutive chunks shows that the processed CSV is ordered by:

1. Date (`QT_INTERVAL_COUNT`)
2. SCATS site (`NB_SCATS_SITE`)
3. Detector (`NB_DETECTOR`)

Site IDs are non-decreasing within the tested chunks, and detector IDs are ordered within sites.

Because chunk boundaries can occur inside a site, feature engineering should not calculate temporal lags independently within each raw CSV chunk.

The processing strategy will therefore preserve temporal history across chunks and process complete dates before generating the final time-series features.

```python
# Temporal feature configuration

LAG_PERIODS = [
    1,      # 15 minutes
    2,      # 30 minutes
    3,      # 45 minutes
    4,      # 60 minutes
    96      # previous day, same 15-minute interval
]

ROLLING_WINDOWS = [
    4,      # previous 1 hour
    8       # previous 2 hours
]

print("Lag periods:", LAG_PERIODS)
print("Rolling windows:", ROLLING_WINDOWS)
```

```python
# Define the final temporal feature names

TEMPORAL_FEATURES = [
    "lag_1",
    "lag_2",
    "lag_3",
    "lag_4",
    "lag_96",
    "rolling_mean_4",
    "rolling_std_4",
    "rolling_mean_8",
    "hour",
    "minute",
    "day_of_week",
    "is_weekend",
    "time_index"
]

TARGET_COLUMN = "target"

print("Temporal features:")
for feature in TEMPORAL_FEATURES:
    print("-", feature)

print("\nTarget:", TARGET_COLUMN)
```

```python
# Convert one day's wide-format traffic data into long format

def reshape_daily_traffic(day_data):
    """
    Convert V00-V95 traffic columns into timestamped long-format data.
    """

    traffic_long = day_data.melt(
        id_vars=[
            "NB_SCATS_SITE",
            "NB_DETECTOR",
            "QT_INTERVAL_COUNT",
            "NM_REGION"
        ],
        value_vars=traffic_columns,
        var_name="interval",
        value_name="volume"
    )

    # Extract interval number from V00, V01, ..., V95
    traffic_long["interval"] = (
        traffic_long["interval"]
        .str.replace("V", "", regex=False)
        .astype("int8")
    )

    # Create the 15-minute time offset
    traffic_long["time"] = pd.to_timedelta(
        traffic_long["interval"] * 15,
        unit="m"
    )

    # Create the actual timestamp
    traffic_long["timestamp"] = (
        traffic_long["QT_INTERVAL_COUNT"]
        + traffic_long["time"]
    )

    return traffic_long
```

```python
# Convert one day's wide-format traffic data into long format

def reshape_daily_traffic(day_data):
    """
    Convert V00-V95 traffic columns into timestamped long-format data.
    """

    traffic_long = day_data.melt(
        id_vars=[
            "NB_SCATS_SITE",
            "NB_DETECTOR",
            "QT_INTERVAL_COUNT",
            "NM_REGION"
        ],
        value_vars=traffic_columns,
        var_name="interval",
        value_name="volume"
    )

    # Extract interval number from V00, V01, ..., V95
    # int16 is used to avoid overflow during minute calculation.
    traffic_long["interval"] = (
        traffic_long["interval"]
        .str.replace("V", "", regex=False)
        .astype("int16")
    )

    # Convert interval number into minutes after midnight
    traffic_long["minutes_after_midnight"] = (
        traffic_long["interval"] * 15
    )

    # Create the 15-minute time offset
    traffic_long["time"] = pd.to_timedelta(
        traffic_long["minutes_after_midnight"],
        unit="m"
    )

    # Create the actual timestamp
    traffic_long["timestamp"] = (
        traffic_long["QT_INTERVAL_COUNT"]
        + traffic_long["time"]
    )

    return traffic_long
```

```python
# Create calendar and time-of-day features

def add_time_features(data):
    """
    Add calendar and time-of-day features to timestamped traffic data.
    """

    data = data.copy()

    data["hour"] = data["timestamp"].dt.hour

    data["minute"] = data["timestamp"].dt.minute

    data["day_of_week"] = data["timestamp"].dt.dayofweek

    data["is_weekend"] = (
        data["day_of_week"] >= 5
    ).astype("int8")

    data["time_index"] = (
        data["hour"] * 4
        + data["minute"] // 15
    ).astype("int8")

    return data
```

```python
# Create short-term lag features

def add_short_term_lags(data):
    """
    Add previous 15, 30, 45, and 60 minute traffic volumes.
    """

    data = data.sort_values(
        [
            "NB_SCATS_SITE",
            "NB_DETECTOR",
            "timestamp"
        ]
    ).copy()

    group_columns = [
        "NB_SCATS_SITE",
        "NB_DETECTOR"
    ]

    for lag in [1, 2, 3, 4]:
        data[f"lag_{lag}"] = (
            data
            .groupby(group_columns)["volume"]
            .shift(lag)
        )

    return data
```

```python
# Validate lag_1 to lag_4 at the Day 2 boundary

boundary = day_2_with_lags[
    day_2_with_lags["timestamp"]
    == pd.Timestamp("2025-04-02 00:00:00")
].copy()

# Create expected values using timestamp-based lookup
expected_boundary = history_buffer[
    history_buffer["timestamp"].isin(
        [
            pd.Timestamp("2025-04-01 23:45:00"),
            pd.Timestamp("2025-04-01 23:30:00"),
            pd.Timestamp("2025-04-01 23:15:00"),
            pd.Timestamp("2025-04-01 23:00:00")
        ]
    )
].copy()

# Pivot previous observations into lag columns
expected_boundary["lag_number"] = (
    (
        pd.Timestamp("2025-04-02 00:00:00")
        - expected_boundary["timestamp"]
    ).dt.total_seconds() // 900
).astype(int)

expected_boundary = expected_boundary.pivot(
    index=[
        "NB_SCATS_SITE",
        "NB_DETECTOR"
    ],
    columns="lag_number",
    values="volume"
).reset_index()

expected_boundary = expected_boundary.rename(
    columns={
        1: "expected_lag_1",
        2: "expected_lag_2",
        3: "expected_lag_3",
        4: "expected_lag_4"
    }
)

# Merge expected values with actual values
validation = boundary.merge(
    expected_boundary,
    on=[
        "NB_SCATS_SITE",
        "NB_DETECTOR"
    ],
    how="left"
)

for lag in [1, 2, 3, 4]:

    mismatch = (
        ~np.isclose(
            validation[f"lag_{lag}"],
            validation[f"expected_lag_{lag}"],
            equal_nan=True
        )
    ).sum()

    print(
        f"lag_{lag} boundary mismatches:",
        mismatch
    )
```

```python
# Create historical rolling features

def add_rolling_features(data):
    """
    Create rolling traffic statistics using only previous observations.
    """

    data = data.sort_values(
        [
            "NB_SCATS_SITE",
            "NB_DETECTOR",
            "timestamp"
        ]
    ).copy()

    group_columns = [
        "NB_SCATS_SITE",
        "NB_DETECTOR"
    ]

    grouped_volume = (
        data
        .groupby(group_columns)["volume"]
    )

    data["rolling_mean_4"] = (
        grouped_volume
        .shift(1)
        .groupby(
            [
                data["NB_SCATS_SITE"],
                data["NB_DETECTOR"]
            ]
        )
        .rolling(4)
        .mean()
        .reset_index(
            level=[0, 1],
            drop=True
        )
    )

    data["rolling_std_4"] = (
        grouped_volume
        .shift(1)
        .groupby(
            [
                data["NB_SCATS_SITE"],
                data["NB_DETECTOR"]
            ]
        )
        .rolling(4)
        .std()
        .reset_index(
            level=[0, 1],
            drop=True
        )
    )

    data["rolling_mean_8"] = (
        grouped_volume
        .shift(1)
        .groupby(
            [
                data["NB_SCATS_SITE"],
                data["NB_DETECTOR"]
            ]
        )
        .rolling(8)
        .mean()
        .reset_index(
            level=[0, 1],
            drop=True
        )
    )

    return data
```

```python
# Iterate through the raw CSV while keeping complete dates together

def iter_complete_days(
    file_path,
    usecols,
    chunksize=50_000
):
    """
    Read the raw CSV in chunks while yielding complete dates.

    A date may span multiple CSV chunks, so the current date
    is carried forward until all of its rows have been read.
    """

    reader = pd.read_csv(
        file_path,
        usecols=usecols,
        chunksize=chunksize
    )

    pending_day = None

    for chunk in reader:

        chunk["QT_INTERVAL_COUNT"] = pd.to_datetime(
            chunk["QT_INTERVAL_COUNT"]
        )

        chunk["date"] = (
            chunk["QT_INTERVAL_COUNT"]
            .dt.normalize()
        )

        dates = chunk["date"].drop_duplicates().tolist()

        for current_date in dates:

            current_part = chunk[
                chunk["date"] == current_date
            ].copy()

            if pending_day is None:

                pending_day = current_part

            else:

                pending_day = pd.concat(
                    [
                        pending_day,
                        current_part
                    ],
                    ignore_index=True
                )

            # If this is not the last date in the chunk,
            # the current date is complete within this chunk.
            if current_date != dates[-1]:

                yield pending_day.drop(
                    columns="date"
                )

                pending_day = None

    # Yield the final pending date
    if pending_day is not None:

        yield pending_day.drop(
            columns="date"
        )
```

```python
# Split a complete day's raw data into manageable batches

def iter_day_batches(
    day_data,
    batch_size=FEATURE_BATCH_SIZE
):
    """
    Yield smaller batches from one complete day's raw data.
    """

    total_rows = len(day_data)

    for start in range(
        0,
        total_rows,
        batch_size
    ):
        end = min(
            start + batch_size,
            total_rows
        )

        yield day_data.iloc[start:end].copy()
```

```python
# Add previous-day traffic values as lag_96

def add_lag_96_to_day(
    current_day,
    previous_day
):
    """
    Add previous-day same-interval traffic as lag_96.

    The current and previous day are matched using:
    SCATS site + detector.
    """

    current = current_day.copy()

    previous = prepare_compact_history(
        previous_day
    )

    # Keep only the columns needed for the join
    previous = previous[
        [
            "NB_SCATS_SITE",
            "NB_DETECTOR"
        ] + traffic_columns
    ]

    # Rename previous-day interval columns
    # so they can be distinguished after the merge.
    previous = previous.rename(
        columns={
            column: f"lag96_{column}"
            for column in traffic_columns
        }
    )

    current = current.merge(
        previous,
        on=[
            "NB_SCATS_SITE",
            "NB_DETECTOR"
        ],
        how="left",
        validate="one_to_one"
    )

    return current
```

```python
# Convert current-day intervals into long format
# while carrying the matching previous-day interval as lag_96

def reshape_with_lag_96(data):
    """
    Convert current-day wide traffic data to long format
    and attach the previous-day same-interval traffic as lag_96.
    """

    current_rows = []

    for interval in traffic_columns:

        temp = data[
            [
                "NB_SCATS_SITE",
                "NB_DETECTOR",
                interval,
                f"lag96_{interval}"
            ]
        ].copy()

        temp["interval"] = interval

        temp = temp.rename(
            columns={
                interval: "volume",
                f"lag96_{interval}": "lag_96"
            }
        )

        current_rows.append(
            temp
        )

    result = pd.concat(
        current_rows,
        ignore_index=True
    )

    return result
```

```python
# Process one manageable batch with lag_96

def process_feature_batch(
    current_batch,
    previous_day
):
    """
    Process one small batch of site-detector records.

    The batch is reshaped into long format and receives:
    - calendar features
    - lag_1 to lag_4
    - rolling features
    - lag_96
    - next 15-minute target within the batch
    """

    # Add previous-day interval values
    batch_with_lag96 = add_lag_96_to_day(
        current_batch,
        previous_day
    )

    # Convert current intervals to long format
    data = reshape_with_lag_96(
        batch_with_lag96
    )

    # Convert interval names into time information
    data["interval_index"] = (
        data["interval"]
        .str.replace("V", "", regex=False)
        .astype("int8")
    )

    data["timestamp"] = (
        pd.to_datetime(
            current_batch["QT_INTERVAL_COUNT"]
            .iloc[0]
        )
        .normalize()
        +
        pd.to_timedelta(
            data["interval_index"] * 15,
            unit="m"
        )
    )

    # Calendar features
    data = add_time_features(
        data
    )

    # Short-term lags
    data = add_short_term_lags(
        data
    )

    # Rolling features
    data = add_rolling_features(
        data
    )

    # Target
    data = add_target(
        data
    )

    return data
```

```python
# Prepare next-day V00 for the current day's final target

def prepare_next_day_target(next_day):
    """
    Prepare next-day V00 values used as the target
    for the previous day's V95.
    """

    target_data = next_day[
        [
            "NB_SCATS_SITE",
            "NB_DETECTOR",
            "V00"
        ]
    ].copy()

    target_data = target_data.rename(
        columns={
            "V00": "next_day_v00"
        }
    )

    return target_data
```

```python
# Apply the next-day target to V95 observations

def apply_next_day_target(
    batch_features,
    next_day_target
):
    """
    Replace the target for 23:45 observations with
    the next day's 00:00 volume.
    """

    data = batch_features.copy()

    data = data.merge(
        next_day_target,
        on=[
            "NB_SCATS_SITE",
            "NB_DETECTOR"
        ],
        how="left",
        validate="many_to_one"
    )

    # V95 corresponds to time_index 95
    final_interval = (
        data["time_index"] == 95
    )

    data.loc[
        final_interval,
        "target"
    ] = data.loc[
        final_interval,
        "next_day_v00"
    ]

    data = data.drop(
        columns=["next_day_v00"]
    )

    return data
```

```python
# Finalize one engineered batch

def finalize_feature_batch(
    current_batch,
    previous_day,
    next_day_target
):
    """
    Create the final modeling dataset for one raw batch.

    Steps:
    1. Add lag_96 from previous day
    2. Reshape current-day intervals
    3. Add calendar features
    4. Add short-term lags
    5. Add rolling features
    6. Add target
    7. Replace V95 target using next-day V00
    8. Keep only final modeling columns
    """

    # -----------------------------------------
    # Step 1: Add previous-day lag_96
    # -----------------------------------------

    data = add_lag_96_to_day(
        current_batch,
        previous_day
    )

    # -----------------------------------------
    # Step 2: Reshape
    # -----------------------------------------

    data = reshape_with_lag_96(
        data
    )

    # -----------------------------------------
    # Step 3: Add interval index
    # -----------------------------------------

    data["interval_index"] = (
        data["interval"]
        .str.replace("V", "", regex=False)
        .astype("int8")
    )

    # -----------------------------------------
    # Step 4: Create timestamp
    # -----------------------------------------

    current_date = pd.to_datetime(
        current_batch["QT_INTERVAL_COUNT"].iloc[0]
    ).normalize()

    data["timestamp"] = (
        current_date
        +
        pd.to_timedelta(
            data["interval_index"] * 15,
            unit="m"
        )
    )

    # -----------------------------------------
    # Step 5: Calendar features
    # -----------------------------------------

    data = add_time_features(
        data
    )

    # -----------------------------------------
    # Step 6: Short-term lags
    # -----------------------------------------

    data = add_short_term_lags(
        data
    )

    # -----------------------------------------
    # Step 7: Rolling features
    # -----------------------------------------

    data = add_rolling_features(
        data
    )

    # -----------------------------------------
    # Step 8: Normal target
    # -----------------------------------------

    data = add_target(
        data
    )

    # -----------------------------------------
    # Step 9: V95 → next-day V00
    # -----------------------------------------

    data = apply_next_day_target(
        data,
        next_day_target
    )

    # -----------------------------------------
    # Step 10: Keep final columns
    # -----------------------------------------

    data = data[
        PRODUCTION_FINAL_COLUMNS
    ]

    return data
```

```python
# Prepare the three-day production simulation

simulation_previous_day = first_day
simulation_current_day = second_day
simulation_next_day = third_day

simulation_next_target = prepare_next_day_target(
    simulation_next_day
)

print("Previous day:")
print(
    simulation_previous_day[
        "QT_INTERVAL_COUNT"
    ].dt.date.unique()
)

print("\nCurrent day:")
print(
    simulation_current_day[
        "QT_INTERVAL_COUNT"
    ].dt.date.unique()
)

print("\nNext day:")
print(
    simulation_next_day[
        "QT_INTERVAL_COUNT"
    ].dt.date.unique()
)
```

```python
# Process one complete day in production-style batches

simulation_output_dir = (
    FEATURED_DATA_DIR
    / "simulation"
)

simulation_output_dir.mkdir(
    parents=True,
    exist_ok=True
)

simulation_part_number = 0

simulation_total_rows = 0

for batch_number, raw_batch in enumerate(
    iter_day_batches(
        simulation_current_day
    ),
    start=1
):

    # -----------------------------------------
    # Feature engineering
    # -----------------------------------------

    feature_batch = finalize_feature_batch(
        raw_batch,
        simulation_previous_day,
        simulation_next_target
    )

    # -----------------------------------------
    # Remove incomplete rows
    # -----------------------------------------

    clean_batch = clean_final_batch(
        feature_batch
    )

    # -----------------------------------------
    # Skip empty batches
    # -----------------------------------------

    if clean_batch.empty:
        continue

    # -----------------------------------------
    # Write to Parquet
    # -----------------------------------------

    simulation_part_number += 1

    output_path = write_feature_batch(
        clean_batch,
        simulation_output_dir,
        simulation_part_number
    )

    simulation_total_rows += len(
        clean_batch
    )

    print(
        f"Batch {batch_number}: "
        f"{len(clean_batch):,} rows → "
        f"{output_path.name}"
    )

print(
    "\nSimulation complete."
)

print(
    "Total rows written:",
    f"{simulation_total_rows:,}"
)

print(
    "Parquet parts:",
    simulation_part_number
)
```
