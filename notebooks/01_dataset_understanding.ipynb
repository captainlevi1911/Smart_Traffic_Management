## Data Quality Check

```python
missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": df.isnull().mean() * 100
})

missing = missing.sort_values(
    "missing_count",
    ascending=False
)

missing[missing["missing_count"] > 0]
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
unique_counts = (
    df.nunique(dropna=False)
      .sort_values()
)

unique_counts
```

```python
print("Unique SCATS sites:",
      df["NB_SCATS_SITE"].nunique())

print("Unique detectors:",
      df["NB_DETECTOR"].nunique())

print("Unique regions:",
      df["NM_REGION"].nunique())
```

## Traffic Volume Columns

```python
traffic_columns = [
    col for col in df.columns
    if col.startswith("V") and col[1:].isdigit()
]

traffic_columns = sorted(
    traffic_columns,
    key=lambda x: int(x[1:])
)

print("Number of traffic columns:", len(traffic_columns))
print(traffic_columns)
```

```python
df[traffic_columns].head()
```

```python
df[traffic_columns].describe().T
```

```python
df[traffic_columns].dtypes.value_counts()
```

```python
negative_values = (df[traffic_columns] < 0).sum()

negative_values[negative_values > 0]
```

```python
zero_values = (df[traffic_columns] == 0).sum()

zero_values.sort_values(ascending=False)
```

```python
interval_sum = df[traffic_columns].sum(axis=1)

comparison = pd.DataFrame({
    "interval_sum": interval_sum,
    "QT_VOLUME_24HOUR": df["QT_VOLUME_24HOUR"]
})

comparison["difference"] = (
    comparison["interval_sum"]
    - comparison["QT_VOLUME_24HOUR"]
)

comparison.head()
```

```python
print(
    "Rows with different values:",
    (comparison["difference"] != 0).sum()
)

print(
    "Maximum absolute difference:",
    comparison["difference"].abs().max()
)
```

## Initial Observations

- NB_SCATS_SITE and NB_DETECTOR appear to identify traffic infrastructure.
- NM_REGION contains regional information.
- QT_VOLUME_24HOUR is an aggregate traffic-related field.
